Self-Attention 계산하고 Q,K,V 가 연산되는 순서를 확인해보기




```python
import numpy as np

# 입력값 정의
Q = np.array([1, 0])
K = np.array([[1, 0],
              [0, 1],
              [1, 1]])
V = np.array([[10, 0],
              [0, 10],
              [5, 5]])

# 1. Q와 K의 내적으로 유사도 점수 계산
scores = np.array([np.dot(Q, k) for k in K])
print("1. 유사도 점수 (Q·K):", scores)

# 2. Softmax 함수 직접 정의
def softmax(x):
    exp_x = np.exp(x - np.max(x))  # 오버플로 방지를 위해 최대값을 빼줌
    return exp_x / np.sum(exp_x)

attention_weights = softmax(scores)
print("2. Attention Weight (softmax 결과):", attention_weights)

# 3. Attention weight를 사용해 Value의 가중합 계산
output = np.zeros_like(V[0], dtype=float)
for weight, v in zip(attention_weights, V):
    output += weight * v
print("3. 최종 출력 (Value 가중합):", output)

# 4. 각 단계 결과 정리 출력
print("\n=== 계산 결과 요약 ===")
print("Q:", Q)
print("K:\n", K)
print("V:\n", V)
print("유사도 점수:", scores)
print("Attention Weight:", attention_weights)
print("최종 Attention 출력:", output)
```

최종 출력: V의 첫 번째와 세 번째 벡터에 더 큰 가중치가 실린 가중합
