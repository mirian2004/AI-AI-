1. 역할 부여 (Role Assignment)

AI에게 특정 역할을 부여하면 해당 관점에서 답변한다

```python
너는 10년차 쇼콜라티에 역할이야. 템퍼링 기법에 대해 설명해줘.
```

**2. 요구사항 구체화 (Specification)**

구체적인 조건을 명시하면 원하는 형식의 답변을 받을 수 있다.

```python
가을에 대한 감성문구를 작성하세요.
10대 청소년을 타켓으로,
캘리그라피 스타일로
1~2줄 분량으로 작성해 주세요.
```

**3. 기호 활용 (Structured Format)**

구분자를 사용하면 프롬프트를 구조화할 수 있다.

```python
### [지시 사항]
아래 제공된 회의록 데이터를 분석하여,
"논의된 안건"과 "결정된 사항"을 구분하여 요약하시오.

### [핵심 제약 사항]
1. 발언자 개인의 "의견 대립"이나 "감정적 표현"은 요약에서 "제거"할 것.
2. "실행 항목"이 있다면 반드시 "담당자"와 "기한"을 함께 명시할 것.
3. 아직 결론이 나지 않은 사항은 "보류 안건"으로 별도 분류할 것.

### [입력 데이터]

오늘 회의에서는 1분기 마케팅 예산안에 대해 논의했습니다.
이 과장은 SNS 광고 비중을 늘려야 한다고 강하게 주장했고, 박 대리는 기존 오프라인 채널도 유지해야 한다며 반박했습니다.
논의 끝에 SNS 광고 예산을 20% 증액하기로 결정했고, 김 대리가 다음 주 금요일까지 세부 집행 계획서를 작성하기로 했습니다.
오프라인 채널 예산 조정 문제는 다음 회의에서 다시 다루기로 했습니다.
```

4. 예시 제공 (Few-shot Learning)

예시를 제공하면 AI가 패턴을 학습하여 일관된 형식으로 답변한다.


```python
당신은 수강신청 보조 AI입니다. 입력된 텍스트를 분석하여 추가로 신청해야할 수업의 종류를 구하세요. 단, 모든 수업은 수업 1개당 3학점으로 고정됩니다.

[예시 1]
입력: "A학생은 18학점을 신청할 수 있으며, 교양 9학점, 전공 6학점을 신청했다. A학생은 이번 신청에서 교양이 전공 학점을 초과해서는 안된다."
출력: 전공 3학점

[예시 2]
입력: "B학생은 21학점을 신청할 수 있으며, 일반선택 3학점, 전공 12학점을 신청했다. B학생은 이번 신청에서 일반선택을 5학점 이상 신청해야한다."
출력: 일반선택 3학점, 전공 3학점

[질문]
C학생은 18학점을 신청할 수 있으며, 교양 3학점, 일반선택 3학점, 전공 6학점을 신청했다. C학생은 이번 신청에서 전공을 가장 많이 들어야하며, 교양,일반선택,전공은 총 학점이 같아서는 안된다.
```
