```python
# [1] 라이브러리 설치 (최초 1회 실행)
!pip install google-generativeai

# [2] 클라이언트 설정
import os
from google import genai
from google.colab import userdata

# API 키 설정
os.environ["GEMINI_API_KEY"] = userdata.get('gemini_api_key')

# API KEY로 클라이언트 설
client = genai.Client()
```

```python
from google.genai import types
def get_ai_response(user_input, temperature=0.7):
  """
  사용자 입력을 받아 Gemini 응답을 반환하는 함수
  """

  # 1. [전처리/유효성 검사] 입력값이 비어있는지 확인
  if not user_input or user_input.strip() == "":
    return "오류: 질문 내용을 입력해주세요."

  # 2. [전처리/프롬프트] 시스템 프롬프트(페르소나) 설정
  # Gemini에서는 시스템 프롬프트를 config에 따로 설정하는 것이 좋음
  system_prompt = "너는 친절하고 명확하게 설명하는 AI 어시스턴트야. 답변은 한국어로 해줘."

  try:
      # 3. [API 호출] 설정된 파라미터와 함께 요청 전송(Gemini 방식)
      response = client.models.generate_content(
          model = 'gemini-flash-latest',
          contents=user_input,
          config=types.GenerateContentConfig(
              system_instruction=system_prompt,
              temperature=temperature,
              max_output_tokens=5000,
          )
      )

      # 4. [후처리/파싱] Gemini 응답에서 텍스트 추출
      ai_text = response.text

      # 5. [후처리/필터링] 금지어 필터링 (안전장)
      banned_words = ["바보", "멍청이"]
      for word in banned_words:
          if word in ai_text:
              return "알림: 부적절한 단어가 포함되어 답변을 표시할 수 없습니다."

      return ai_text

  except Exception as e:
      # 6. [예외처리] API 호출 실패 시 대응
      return f"시스템 오류 발생: {str(e)}"
```

```python
# [테스트 1] 일반적인 질문
question = "API가 무엇인지 초등학생도 이해하게 설명해줘."
print(f"Q: {question}")
answer = get_ai_response(question)
print("-" * 50)
print(f"A: {answer}")

print("\n" + "=" * 50 + "\n")

# [테스트 2] 창의성이 필요한 질문 (Temperature 높게)
question_creative = "기발한 SF 영화 줄거리 하나만 짧게 지어줘."
print(f"Q: {question_creative}")

# temperature = 1.2로 설정하여 더 창의적인 답변 유도
answer_creative = get_ai_response(question_creative, temperature=1.2)
print("-" * 50)
print(f"A: {answer_creative}")

print("\n" + "=" * 50 + "\n")

# [테스트 3] 빈 값 입력 (유효성 검사 테스트)
print("Q: (빈 칸)")
error_test = get_ai_response("")
print("-" * 50)
print(f"A: {error_test}")
```
