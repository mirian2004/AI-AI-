```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. 데이터 준비 (공부 시간 -> 시험 점수)
# 정답 데이터 (Reality)
x_train = np.array([1,2,3,4,5]) #공부 시간
y_true = np.array([10,20,30,40,50]) #실제 점수 (규칙: 점수 = 10 * 시간)

print(f"학습 데이터: {x_train}")
print(f"학습 데이터: {y_true}")

plt.figure(figsize=(10,5))
plt.scatter(x_train, y_true, color='blue', label='Training Data')
plt.plot(x_train, y_true, color='red', linestyle='--',linewidth=2, label='True Regression')
plt.title('Training Data Visualization')
plt.xlabel('Study Time (Hours)')
plt.ylabel('Test Score (Points)')
plt.legend()
plt.show()
```

```python
# 2. 초기 AI 모델 설정 (무작위 상태)
# y = w * x + b (직선 공식)
w = np.random.rand() #가중치
b = np.random.rand() #편향
learning_rate = 0.01 #학습률

epochs = 50 #반복횟수
loss_history = [] #Loss 변화를 기록할 리스트

print(f"\n 학습 시작! (초기 w: {w:.2f}, b: {b:.2f})")
```

```python
# 3. 반복 학습 (Epoch) 진행
for epoch in range(epochs):

  #[1단계] 예측하기
  y_pred = w*x_train + b

  #[2단계] Loss 계산 - MSE
  #MSE = (예측값 - 정답)^2 의 평균
  loss = np.mean((y_pred - y_true)** 2)

  loss_history.append(loss)

  #[3단계] Gradient Descent - 미분
  #(Loss를 줄이는 방향으로 w 와 b 를 약간씩 수정)
  w_grad = -2 * np.mean(x_train*(y_true-y_pred))
  b_grad = -2 * np.mean(y_true-y_pred)

  w -= learning_rate * w_grad #가중치 업데이트
  b -= learning_rate * b_grad #편향 업데이트

  #10번마다 로그 출력
  if(epoch +1) % 10 == 0:
    print(f"[Epoch {epoch+1}/{epochs}] Loss: {loss:.4f} (예측: {y_pred[0]:.1f}, 정답: {y_true[0]})")

print(f"학습종료! (최종 w: {w:.2f}, b: {b:.2f})")
```

```python
# 4. [주요 아웃풋] Loss 감소 그래프 시각화
plt.figure(figsize=(10,5))
plt.plot(range(1,epochs+1), loss_history, label='Training Loss', color='red', linewidth=2)
plt.title('Loss Reduction Curve (Learning Process)')
plt.xlabel('Epochs (Study Count)')
plt.ylabel('Loss (MSE Error)')
plt.grid(True)
plt.legend()
plt.show()
```

결과 : 가중치가 10에 가깝고 편향은 2 초반대로 나오게 학습되었다.
