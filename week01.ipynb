```python
# 필요한 라이브러리 불러오기
import torch
import torch.nn as nn
from torchsummary import summary
```

```python
# 신경망 모델 정의
class SimpleNN(nn.Module):
  def __init__(self):
    super(SimpleNN, self).__init__()

    # 1024개의 입력 특성
    # 첫 번쨰 은닉층: 1024 차원 -> 256 차원
    # fc(Fully Connected Layer): 한 층의 모든 뉴런이 다음층의 모든 뉴런과 연결된 형태.
    self.fc1 = nn.Linear(1024, 256)
    self.relu1 = nn.ReLU()

    # 두 번째 은닉층 : 256 -> 128
    self.fc2 = nn.Linear(256, 128)
    self.relu2 = nn.ReLU()

    # 세 번째 은닉층 : 128 -> 64
    self.fc3 = nn.Linear(128, 64)
    self.relu3 = nn.ReLU()

    # 출력층 : 64 -> 10 (10개 클래스로 분류)
    self.fc4 = nn.Linear(64, 10)


  # forward : 입력 데이터가 어떤 레이어를 순서대로 통과하는지 정의
  # 이 함수의 코드 흐름이 신경망의 연산 흐름과 같다.
  # 이 부분을 보면 입력 데이터가 어떻게 변형되어 최종 출력으로 이어지는지 한눈에 파악가능
  def forward(self, x):
      x = self.fc1(x)
      x = self.relu1(x)
      x = self.fc2(x)
      x = self.relu2(x)
      x = self.fc3(x)
      x = self.relu3(x)
      x = self.fc4(x)
      return x
```

```python
# 모델 인스턴스 생성
model = SimpleNN()

# 모델 요약 출력
print("Neural Network Architecture Summary:")

# summary 함수에는 임의 입력이 모양(shape)을 입력해야한다
# 위 신경망에서 1024개의 특성을 가진 입력을 가정했으므로 -> (데이터개수, 1024)
summary(model, input_size = (1,1024))
```
